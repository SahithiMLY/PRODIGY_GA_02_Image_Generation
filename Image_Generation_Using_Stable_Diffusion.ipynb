```python
import keras_cv

model = keras_cv.models.StableDiffusion(
    img_width=512,
    img_height=512
)
```

```python
prompt = "A futuristic city with flying cars at sunset"

images = model.text_to_image(
    prompt,
    batch_size=1
)

```

```python
import matplotlib.pyplot as plt

plt.imshow(images[0])
plt.axis("off")
plt.show()
```
